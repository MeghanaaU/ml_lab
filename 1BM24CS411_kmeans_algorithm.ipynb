```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Load dataset
data = pd.read_csv("iris(2).csv")

# Use only numerical features (ignore label column if present)
X = data.select_dtypes(include=['float64', 'int64'])

# Optional: Scale features (recommended for K-Means)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply K-Means (k = 3 for iris dataset)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(X_scaled)

# Cluster labels
data['Cluster'] = kmeans.labels_

# Print results
print("Cluster Centers:\n", kmeans.cluster_centers_)
print("\nFirst 10 rows with cluster labels:\n")
print(data.head(10))
```
